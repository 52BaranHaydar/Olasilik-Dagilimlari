## 📌 1. Giriş ve Z-Dönüşümü (Standardizasyon)

**Standart Normal Dağılım (Standard Normal Distribution)**, ortalaması sıfır ($\mu = 0$) ve varyansı bir ($\sigma^2 = 1, \sigma = 1$) olan özel bir Normal Dağılım halidir ve genellikle $Z \sim \mathcal{N}(0, 1)$ olarak sembolize edilir.

### ❓ Neden Standardizasyon (Z-Skoru) Yaparız?
1. **Farklı Ölçekleri Karşılaştırabilme:** Farklı birimlerde ölçülen (örneğin boy: cm, kilo: kg, maaş: TL, yaş: yıl) verileri aynı ortak ölçekte ($Z$-ekseni) kıyaslanabilir hale getirir.
2. **Olasılık ve Yüzdelik Dilim Hesaplama:** Herhangi bir $\mathcal{N}(\mu, \sigma^2)$ dağılımını $Z$-tablosu veya standart fonksiyonlar ile kolayca analiz etmemizi sağlar.
3. **Makine Öğrenmesi Algoritmaları:** Gradient Descent, KNN, SVM, PCA ve Yapay Sinir Ağları gibi algoritmalar değişkenlerin aynı ölçekte ($Z \sim \mathcal{N}(0, 1)$) olmasına ihtiyaç duyar.
4. **Anomali ve Aykırı Değer Tespiti:** Bir gözlemin ortalamadan kaç standart sapma uzakta olduğunu doğrudan gösterir.

---
## 📐 2. Matematiksel Formülasyon ve Simetri Kuralları

### 🔹 Z-Dönüşümü (Z-Score) Formülü:
Herhangi bir $X \sim \mathcal{N}(\mu, \sigma^2)$ normal dağılımından gelen rastgele değişken için:
$$Z = \frac{X - \mu}{\sigma}$$

- $Z = 0 \implies X = \mu$ (Değer tam ortalamada)
- $Z = +1.5 \implies X$, ortalamanın $1.5\sigma$ **üzerinde**
- $Z = -2.0 \implies X$, ortalamanın $2.0\sigma$ **altında**

---

### 🔹 Standart Olasılık Yoğunluk Fonksiyonu $\phi(z)$:
$$\phi(z) = \frac{1}{\sqrt{2\pi}} e^{-\frac{z^2}{2}}, \quad z \in (-\infty, +\infty)$$

### 🔹 Standart Kümülatif Dağılım Fonksiyonu $\Phi(z)$:
$$\Phi(z) = P(Z \le z) = \int_{-\infty}^{z} \frac{1}{\sqrt{2\pi}} e^{-\frac{t^2}{2}} dt$$

---

### 🔹 Simetri ve Kritik Olasılık Kuralları:
1. **Negatif Simetri:** $\Phi(-z) = 1 - \Phi(z) \implies P(Z \le -z) = P(Z \ge z)$
2. **Simetrik Aralık Olasılığı:** $P(-z \le Z \le z) = \Phi(z) - \Phi(-z) = 2\Phi(z) - 1$
3. **Aralık Olasılığı:** $P(a \le Z \le b) = \Phi(b) - \Phi(a)$

## 💻 3. Python Ortam Hazırlığı ve Z-Tablosu (Z-Table) Üretimi

```python
from IPython.display import display
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import norm

# Grafik tema ayarları
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['figure.dpi'] = 110

print("✅ Kütüphaneler başarıyla yüklendi.")
```

### 📊 Klasik Z-Tablosunun (Standard Normal Cumulative Table) Python ile Oluşturulması
İstatistik ders kitaplarının arkasında yer alan meşhur **Z-Tablosunu** Pandas matrisi olarak oluşturalım ($Z \in [0.0, 3.0]$):

```python
# Z-Tablosu: Satırlar 0.0'dan 3.0'a (0.1 adımlarla), Sütunlar 0.00'dan 0.09'a (0.01 adımlarla)
z_rows = np.arange(0.0, 3.1, 0.1)
z_cols = np.arange(0.00, 0.10, 0.01)

z_table_data = []
for r in z_rows:
    row_vals = [norm.cdf(round(r + c, 2)) for c in z_cols]
    z_table_data.append(row_vals)

df_z_table = pd.DataFrame(z_table_data, index=[f"{r:.1f}" for r in z_rows], columns=[f"{c:.2f}" for c in z_cols])

def lookup_z_table(z_val):
    z_rounded = round(z_val, 2)
    row_key = f"{np.floor(z_rounded * 10) / 10:.1f}"
    col_key = f"{z_rounded - float(row_key):.2f}"
    prob = norm.cdf(z_val)
    print(f"🔍 Z = {z_val:.2f} için Tablo Değeri -> Satır: {row_key}, Sütun: {col_key} => Φ({z_val:.2f}) = {prob:.4f} (%{prob*100:.2f})")
    return prob

print("📑 Standart Z-Tablosu (İlk 8 Satır):")
display(df_z_table.head(8))

lookup_z_table(1.96)
lookup_z_table(1.645)
```

## 🔬 4. Z-Skoru ile Aykırı Değer (Outlier) ve Anomali Tespiti

Veri biliminde aykırı değerleri belirlemek için iki temel Z-Skoru yaklaşımı kullanılır:

1. **Standart Z-Score Yöntemi ($|Z| > 3$ Kuralı):**
   - Değer $\mu \pm 3\sigma$ aralığının dışındaysa aykırı değer (outlier) kabul edilir.
   - *Dezavantajı:* Ortalama ve standart sapma aşırı uç değerlerden etkilenir.

2. **Dayanıklı (Robust / Modified) Z-Score Yöntemi (Medyan & MAD):**
   - Ortalama yerine **Medyan**, Standart Sapma yerine **MAD (Median Absolute Deviation)** kullanılır:
   $$\text{MAD} = \text{median}(|X_i - \text{median}(X)|)$$
   $$\text{Modified } Z_i = \frac{0.6745 \cdot (X_i - \text{median}(X))}{\text{MAD}}$$
   - $|\text{Modified } Z| > 3.5$ ise aykırı değer olarak etiketlenir.

```python
np.random.seed(42)
# %98'i normal dağılan, %2'si ekstrem anomali içeren sentetik veri
veri = np.random.normal(loc=100, scale=10, size=200)
veri[15] = 185  # Ekstrem yüksek anomali
veri[88] = 20   # Ekstrem düşük anomali

# 1. Standart Z-Score
z_std = (veri - np.mean(veri)) / np.std(veri)
outliers_std = np.where(np.abs(z_std) > 3)[0]

# 2. Modified (Robust) Z-Score
medyan = np.median(veri)
mad = np.median(np.abs(veri - medyan))
modified_z = 0.6745 * (veri - medyan) / mad
outliers_robust = np.where(np.abs(modified_z) > 3.5)[0]

print(f"📊 Standart Z-Score ile Tespit Edilen Aykırı İndeksler : {outliers_std} -> Değerler: {veri[outliers_std]}")
print(f"🛡️ Robust Z-Score ile Tespit Edilen Aykırı İndeksler   : {outliers_robust} -> Değerler: {veri[outliers_robust]}")
```

## 📈 5. Kapsamlı ve Estetik Görselleştirmeler

### 🎨 Grafik 1: Kritik Z Değerleri ve Anlamlılık Düzeyleri ($\alpha = 0.05, 0.01, 0.001$)

```python
z_axis = np.linspace(-4, 4, 1000)
phi_axis = norm.pdf(z_axis, 0, 1)

plt.figure(figsize=(12, 6))
plt.plot(z_axis, phi_axis, color='#2c3e50', linewidth=2.5, label='Standard Normal N(0, 1)')

# %95 Güven Aralığı (Z = ±1.96, alpha = 0.05)
z_95 = 1.96
mask_95 = (z_axis >= -z_95) & (z_axis <= z_95)
plt.fill_between(z_axis[mask_95], phi_axis[mask_95], color='#2ecc71', alpha=0.4, label='Orta %95 Alan (|Z| ≤ 1.96)')

# %95 Ret / Kuyruk Bölgeleri (Sol ve Sağ %2.5)
mask_tail_left = (z_axis < -z_95)
mask_tail_right = (z_axis > z_95)
plt.fill_between(z_axis[mask_tail_left], phi_axis[mask_tail_left], color='#e74c3c', alpha=0.6, label='Kritik Kuyruk Alanı (%2.5 + %2.5 = %5)')
plt.fill_between(z_axis[mask_tail_right], phi_axis[mask_tail_right], color='#e74c3c', alpha=0.6)

plt.axvline(-1.96, color='#c0392b', linestyle='--', linewidth=2, label='Z = -1.96 (p = 0.025)')
plt.axvline(1.96, color='#c0392b', linestyle='--', linewidth=2, label='Z = +1.96 (p = 0.975)')
plt.axvline(0, color='gray', linestyle=':', linewidth=1.5)

plt.title("Standart Normal Dağılım: Kritik Z Değerleri ve %95 Güven Düzeyi (α = 0.05)", fontsize=13, fontweight='bold', pad=12)
plt.xlabel("Z Skoru", fontsize=11)
plt.ylabel("Olasılık Yoğunluğu φ(z)", fontsize=11)
plt.legend(loc='upper right', frameon=True, facecolor='white', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 🎨 Grafik 2: Ham Veri vs Z-Standardize Edilmiş Veri Karşılaştırması

```python
np.random.seed(42)
ham_veri = np.random.normal(loc=175, scale=12, size=1000) # Örneğin boy verisi (cm)
standard_veri = (ham_veri - np.mean(ham_veri)) / np.std(ham_veri) # Z-skoru

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Ham Veri Dağılımı
sns.histplot(ham_veri, bins=30, kde=True, ax=ax1, color='#3498db', edgecolor='black')
ax1.axvline(np.mean(ham_veri), color='#e74c3c', linestyle='--', linewidth=2, label=f'μ = {np.mean(ham_veri):.1f} cm')
ax1.set_title("Ham Veri Dağılımı: Boy (cm)\nμ = 175, σ = 12", fontsize=12, fontweight='bold')
ax1.set_xlabel("Boy (cm)")
ax1.set_ylabel("Frekans")
ax1.legend(frameon=True, facecolor='white')
ax1.grid(True, linestyle='--', alpha=0.6)

# Standardize Veri Dağılımı
sns.histplot(standard_veri, bins=30, kde=True, ax=ax2, color='#9b59b6', edgecolor='black')
ax2.axvline(0, color='#e74c3c', linestyle='--', linewidth=2, label='μ = 0')
ax2.set_title("Standardize Edilmiş Veri: Z-Skoru\nμ = 0, σ = 1", fontsize=12, fontweight='bold')
ax2.set_xlabel("Z Değeri (Standart Sapma Birimi)")
ax2.set_ylabel("Frekans")
ax2.legend(frameon=True, facecolor='white')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

## 💼 6. Gerçek Hayat Problemleri ve Vaka Analizleri

### 🎓 Vaka 1: Üniversite Çan Eğrisi Sınav Notlandırması (Adil Başarı Kıyaslaması)
Ahmet iki farklı dersten sınavlara girmiştir:
- **Matematik Sınavı:** Sınıf Ortalaması $\mu_M = 55$, Standart Sapma $\sigma_M = 15$. Ahmet'in notu $X_M = 70$.
- **Fizik Sınavı:** Sınıf Ortalaması $\mu_F = 75$, Standart Sapma $\sigma_F = 8$. Ahmet'in notu $X_F = 80$.

Ahmet ham puan olarak Fizik'ten daha yüksek almış gibi görünse de, sınıfındaki sıralama ve başarısı açısından **hangi derste daha başarılıdır?**

```python
# Matematik Z-Skoru
z_mat = (70 - 55) / 15

# Fizik Z-Skoru
z_fiz = (80 - 75) / 8

# Yüzdelik Dilimler (Percentile Rank)
p_mat = norm.cdf(z_mat) * 100
p_fiz = norm.cdf(z_fiz) * 100

print(f"📐 Matematik Sınavı: Z = {z_mat:+.2f}  -> Sınıfın %{p_mat:.1f}'inden daha başarılı")
print(f"⚡ Fizik Sınavı     : Z = {z_fiz:+.2f}  -> Sınıfın %{p_fiz:.1f}'inden daha başarılı")
print()
if z_mat > z_fiz:
    print("🏆 Sonuç: Ahmet Matematik dersinde sınıfına göre ÇOK DAHA BAŞARILIDIR! (Z_mat > Z_fiz)")
else:
    print("🏆 Sonuç: Ahmet Fizik dersinde sınıfına göre daha başarılıdır.")
```

### 🛡️ Vaka 2: Siber Güvenlik - DDoS Saldırısı Gerçek Zamanlı Ağ Trafiği Anomali Dedektörü
Bir sunucunun normal çalışma saatlerinde gelen ağ trafiği ortalama $\mu = 60\text{ Mbps}$, standart sapması $\sigma = 5\text{ Mbps}$ olan bir normal dağılıma uymaktadır.
- Ağ izleme sistemine aniden $120\text{ Mbps}$ trafik geldiğinde $Z$-skoru ne olur?
- Sistem $|Z| > 3.0$ durumunda otomatik alarm üretiyorsa bu trafik alarm üretir mi?

```python
mu_ag = 60.0    # Mbps
sigma_ag = 5.0   # Mbps
gelen_trafik = 120.0 # Mbps

z_ag = (gelen_trafik - mu_ag) / sigma_ag
p_ddos = norm.sf(z_ag) # P(Z >= 12)

print(f"🌐 Normal Ağ Trafiği    : μ = {mu_ag} Mbps, σ = {sigma_ag} Mbps")
print(f"🚨 Anlık Gözlenen Trafik: {gelen_trafik} Mbps")
print(f"⚡ Hesaplanan Z-Skoru   : Z = {z_ag:+.2f}")
print(f"🎲 Doğal Olarak Olma Olasılığı: {p_ddos:.2e}")

if abs(z_ag) > 3.0:
    print("🔴 ALARM TETİKLENDİ: Trafik 12-Sigma seviyesinde bir anomalidir! Potansiyel DDoS Saldırısı!")
else:
    print("🟢 Normal trafik akışı.")
```

## 📝 7. Özet ve Hızlı Başvuru Kartı

| Özellik | Formül / Değer | Açıklama |
| :--- | :--- | :--- |
| **Ortalama ($\mu$)** | $0$ | Dağılım tam olarak 0 merkezlidir |
| **Standart Sapma ($\sigma$)** | $1$ | Ölçek standart 1 birimdir |
| **Varyans ($\sigma^2$)** | $1$ | $\text{Var}(Z) = 1$ |
| **Z-Dönüşüm Formülü** | $Z = \frac{X - \mu}{\sigma}$ | Herhangi bir normal veriyi standartlaştırır |
| **Geri Dönüşüm (Ters Z)** | $X = \mu + Z \cdot \sigma$ | Z skorunu orijinal veri ölçeğine çevirir |
| **Kritik Z Değerleri** | $\pm 1.645$ (%90), $\pm 1.96$ (%95), $\pm 2.58$ (%99) | Hipotez testleri ve güven aralıkları için temel eşikler |
| **Aykırı Değer Kuralı** | $|Z| > 3$ | 3-Sigma kuralı dışındaki nadir gözlemler |
| **SciPy Karşılığı** | `scipy.stats.norm(0, 1)` | Parametresiz çağrıldığında varsayılan olarak $\mathcal{N}(0, 1)$'dir |
