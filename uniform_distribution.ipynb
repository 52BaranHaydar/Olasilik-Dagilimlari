# 📚 DÜZGÜN (UNIFORM) DAĞILIM - DETAYLI TEORİ VE UYGULAMA REHBERİ
### Olasılık Dağılımları Serisi - Sürekli ve Ayrık Dağılımlar

---

## 📌 1. Giriş ve Sezgisel Açıklama

**Düzgün Dağılım (Uniform Distribution)** veya **Dikdörtgen Dağılım (Rectangular Distribution)**, tanımlı olduğu aralıktaki **tüm olayların gerçekleşme olasılığının birbirine eşit olduğu** en temel ve sezgisel olasılık dağılımıdır.

Düzgün dağılım iki ana kategoride incelenir:
1. **Ayrık Düzgün Dağılım (Discrete Uniform):** Sonlu sayıda ve sayılabilir tam sayı sonuç vardır (Örn: Hilesiz zar atışında $1, 2, 3, 4, 5, 6$ gelme olasılıklarının her biri $\frac{1}{6}$'dır).
2. **Sürekli Düzgün Dağılım (Continuous Uniform):** $[a, b]$ reel sayı aralığındaki sonsuz sayıdaki noktanın her biri eşit olasılık yoğunluğuna sahiptir (Örn: Bir sonraki otobüsün $0$ ile $15$ dakika arasında herhangi bir anda gelmesi).

---
## 📐 2. Matematiksel Tanım ve Olasılık Yoğunluk Fonksiyonu (PDF)

### 🔹 Sürekli Düzgün Dağılım $X \sim \mathcal{U}(a, b)$:
$a$ alt sınır (minimum) ve $b$ üst sınır (maksimum) olmak üzere ($b > a$):

#### 1. Olasılık Yoğunluk Fonksiyonu (PDF):
$$f(x) = \begin{cases} \frac{1}{b - a}, & a \le x \le b \\ 0, & \text{diğer durumlarda} \end{cases}$$

#### 2. Kümülatif Dağılım Fonksiyonu (CDF):
$$F(x) = P(X \le x) = \begin{cases} 0, & x < a \\ \frac{x - a}{b - a}, & a \le x \le b \\ 1, & x > b \end{cases}$$

#### 3. Belirli Bir Aralık Olasılığı:
$[x_1, x_2] \subseteq [a, b]$ aralığı için olasılık doğrudan aralık uzunluğunun toplam aralığa oranıdır:
$$P(x_1 \le X \le x_2) = F(x_2) - F(x_1) = \frac{x_2 - x_1}{b - a}$$

---

### 📊 İstatistiksel Özellikler ve Momentler

| İstatistiksel Metrik | Formül | Açıklama |
| :--- | :--- | :--- |
| **Beklenen Değer (Mean)** | $E[X] = \frac{a + b}{2}$ | Dağılımın tam orta noktası |
| **Medyan (Median)** | $\text{Medyan} = \frac{a + b}{2}$ | Alanı iki eşit parçaya bölen değer |
| **Mod (Mode)** | Yok (Tanımsız) | Tüm değerler eşit olasılık yoğunluğuna sahiptir |
| **Varyans (Variance)** | $\text{Var}(X) = \frac{(b - a)^2}{12}$ | Aralığın karesinin 12'ye bölümü |
| **Standart Sapma** | $\sigma = \frac{b - a}{\sqrt{12}} = \frac{b - a}{2\sqrt{3}}$ | Ortalamadan sapma genişliği |
| **Çarpıklık (Skewness)** | $\gamma_1 = 0$ | Mükemmel simetriktir |
| **Basıklık (Kurtosis)** | $\beta_2 = 1.8 \quad (\text{Excess} = -1.2)$ | Platikurtik (Düz tepeli) |

## 🎯 3. Bakkal Şeker Satış Problemi - Detaylı Analiz ve Matematiksel Çözüm

> **Soru:** Bir bakkalın günlük sattığı şeker miktarı $40\text{ kg}$ ile $60\text{ kg}$ arasında Sürekli Düzgün Dağılım ($X \sim \mathcal{U}(40, 60)$) göstermektedir.
> 1. Bakkalın bir günde $[50\text{ kg}, 55\text{ kg}]$ arasında şeker satma olasılığı nedir?
> 2. Bakkalın bir günde $55\text{ kg}$'dan fazla satma olasılığı nedir?
> 3. Günlük beklenen ortalama satış ve standart sapma nedir?

```python
from IPython.display import display
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import uniform

# Problem Parametreleri
a = 40  # Minimum satış (kg)
b = 60  # Maksimum satış (kg)
aralik = b - a  # scale = 20 kg

# 1. P(50 <= X <= 55)
# Matematiksel Formül: (55 - 50) / (60 - 40) = 5 / 20 = 0.25
p_50_55_scipy = uniform.cdf(55, loc=a, scale=aralik) - uniform.cdf(50, loc=a, scale=aralik)

# 2. P(X > 55)
# Matematiksel Formül: (60 - 55) / (60 - 40) = 5 / 20 = 0.25
p_ust_55 = uniform.sf(55, loc=a, scale=aralik)

# 3. İstatistiksel Momentler
beklenen_satis, varyans, carpiklik, basiklik = uniform.stats(loc=a, scale=aralik, moments='mvsk')
std_satis = np.sqrt(varyans)

print(f"🔹 1. [50 - 55] kg Arasında Satış Olasılığı P(50 <= X <= 55) : %{p_50_55_scipy * 100:.2f} (0.25)")
print(f"🔹 2. 55 kg'dan Fazla Satış Olasılığı P(X > 55)             : %{p_ust_55 * 100:.2f} (0.25)")
print(f"🔹 3. Beklenen Günlük Ortalama Satış E[X]                  : {beklenen_satis:.2f} kg")
print(f"🔹 4. Satış Miktarı Standart Sapması σ                     : {std_satis:.2f} kg (Formül: 20 / √12 = {20/np.sqrt(12):.2f})")
```

### 🎨 Bakkal Probleminin Kusursuz Görselleştirmesi

```python
x_vals = np.linspace(35, 65, 1000)
pdf_vals = uniform.pdf(x_vals, loc=a, scale=aralik)

plt.figure(figsize=(11, 5.5))
plt.plot(x_vals, pdf_vals, color='#2c3e50', linewidth=2.5, label='PDF f(x) = 1 / 20 = 0.05')

# [50 - 55] Aralığını Vurgula
mask_target = (x_vals >= 50) & (x_vals <= 55)
plt.fill_between(x_vals[mask_target], pdf_vals[mask_target], color='#e74c3c', alpha=0.5, 
                 label='Hedef Bölge: P(50 ≤ X ≤ 55) = %25 (Alan = 5 × 0.05 = 0.25)')

# Kalan [40-50] ve [55-60] bölgeleri
mask_other = ((x_vals >= 40) & (x_vals < 50)) | ((x_vals > 55) & (x_vals <= 60))
plt.fill_between(x_vals[mask_other], pdf_vals[mask_other], color='#3498db', alpha=0.25, label='Diğer Geçerli Aralıklar')

plt.axvline(40, color='gray', linestyle='--', label='Alt Sınır a = 40 kg')
plt.axvline(60, color='gray', linestyle='--', label='Üst Sınır b = 60 kg')
plt.axvline(50, color='#e74c3c', linestyle=':')
plt.axvline(55, color='#e74c3c', linestyle=':')

plt.title("Bakkal Şeker Satışı: Düzgün Dağılım U(40, 60) ve [50-55] kg Olasılık Alanı", fontsize=12, fontweight='bold', pad=12)
plt.xlabel("Günlük Şeker Satışı (kg)", fontsize=11)
plt.ylabel("Olasılık Yoğunluğu f(x)", fontsize=11)
plt.ylim(0, 0.075)
plt.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9.5)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 🔬 4. SciPy ile Düzgün Dağılım Fonksiyonları (PDF, CDF, PPF, RVS, Stats)

`scipy.stats.uniform` sınıfı için **ÖNEMLİ KURAL**:
- `loc = a` (Başlangıç / Alt Sınır)
- `scale = b - a` (Aralık Uzunluğu, **üst sınır DEĞİLDİR!**)

Örneğin $U(10, 30)$ dağılımı için: `uniform(loc=10, scale=20)` olarak tanımlanmalıdır.

```python
# U(10, 30) Dağılımı Tanımlama
dist_u = uniform(loc=10, scale=20)

# 1. 25'e kadar olan kümülatif olasılık P(X <= 25)
p_cdf = dist_u.cdf(25)

# 2. Yüzde 70'lik dilim (70. Persentil)
val_p70 = dist_u.ppf(0.70)

# 3. Rastgele 5 Örneklem Üretme
samples = dist_u.rvs(size=5, random_state=42)

print(f"🔹 P(X <= 25)                   : {p_cdf:.4f} (%{p_cdf*100:.1f})")
print(f"🔹 70. Yüzdelik Değeri (P70)    : {val_p70:.2f} (Formül: 10 + 0.70 * 20 = {10 + 0.70*20})")
print(f"🔹 Üretilen Rastgele Örneklemler: {samples.round(2)}")
```

### 🎨 Grafik 3: Düzgün Dağılım ile Monte Carlo Yöntemiyle $\pi$ Sayısının Hesaplanması

Düzgün dağılımın en çarpıcı uygulamalarından biri **Monte Carlo Simülasyonu** ile $\pi$ sayısı tahminidir.
- $[-1, 1] \times [-1, 1]$ karesine rastgele $X, Y \sim \mathcal{U}(-1, 1)$ noktaları atılır.
- Daire içine ($X^2 + Y^2 \le 1$) düşen nokta sayısı / Toplam nokta sayısı $\approx \frac{\pi r^2}{(2r)^2} = \frac{\pi}{4}$
- $\pi \approx 4 \times \frac{\text{Daire İçi Noktalar}}{\text{Toplam Noktalar}}$

```python
np.random.seed(42)
N = 10000

# [-1, 1] aralığında düzgün dağılan X ve Y koordinatları
x_pts = np.random.uniform(-1, 1, N)
y_pts = np.random.uniform(-1, 1, N)

# Çember içi testi: x² + y² <= 1
inside_circle = (x_pts**2 + y_pts**2) <= 1
pi_tahmini = 4 * np.sum(inside_circle) / N

plt.figure(figsize=(7, 7))
plt.scatter(x_pts[inside_circle], y_pts[inside_circle], color='#2ecc71', s=2, alpha=0.6, label='Çember İçi Noktalar')
plt.scatter(x_pts[~inside_circle], y_pts[~inside_circle], color='#e74c3c', s=2, alpha=0.6, label='Çember Dışı Noktalar')

# Referans Çember Çizimi
theta = np.linspace(0, 2*np.pi, 200)
plt.plot(np.cos(theta), np.sin(theta), color='black', linewidth=2, label='Birim Çember (r=1)')

plt.title(f"Monte Carlo ile π Sayısı Tahmini\n{N:,} Nokta ile Tahmin Edilen π ≈ {pi_tahmini:.4f} (Hata: %{abs(pi_tahmini - np.pi)/np.pi*100:.3f})", 
          fontsize=12, fontweight='bold', pad=10)
plt.xlabel("X Koordinatı ~ U(-1, 1)")
plt.ylabel("Y Koordinatı ~ U(-1, 1)")
plt.axis('equal')
plt.legend(loc='upper right', frameon=True, facecolor='white', fontsize=9)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 🎨 Grafik 4: Ters Dönüşüm Yöntemi (Inverse Transform Sampling)

Tüm bilgisayarlar ve programlama dilleri, karmaşık olasılık dağılımlarından (örneğin Üstel Dağılım $\text{Exp}(\lambda)$) rastgele sayı üretmek için temel **$U \sim \mathcal{U}(0, 1)$ Düzgün Dağılımını** kullanır!
$$X = F^{-1}(U) = -\frac{\ln(1 - U)}{\lambda}$$

```python
np.random.seed(42)
n_sim = 5000
lam = 0.5  # Üstel dağılım parametresi lambda

# 1. Adım: Standart Düzgün Dağılımdan Sayı Üret
u_samples = np.random.uniform(0, 1, n_sim)

# 2. Adım: Ters CDF ile Üstel Dağılıma Dönüştür
exp_simulated = -np.log(1 - u_samples) / lam

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Düzgün Girdi
sns.histplot(u_samples, bins=30, ax=ax1, color='#3498db', stat='density', edgecolor='black')
ax1.set_title("1. Adım: Standart Düzgün Dağılım Girdisi\nU ~ U(0, 1)", fontsize=12, fontweight='bold')
ax1.set_xlabel("U")
ax1.grid(True, linestyle='--', alpha=0.6)

# Üretilen Üstel Dağılım Çıktısı
sns.histplot(exp_simulated, bins=30, ax=ax2, color='#e67e22', stat='density', kde=True, edgecolor='black')
x_plot = np.linspace(0, 12, 200)
ax2.plot(x_plot, lam * np.exp(-lam * x_plot), 'r--', linewidth=2, label='Teorik Üstel PDF')
ax2.set_title("2. Adım: Ters Dönüşüm ile Üretilen Dağılım\nX = -ln(1-U)/λ ~ Exp(λ=0.5)", fontsize=12, fontweight='bold')
ax2.set_xlabel("X")
ax2.legend(frameon=True, facecolor='white')
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```

### 🚌 Vaka 1: Toplu Taşıma - Metro/Otobüs Bekleme Süresi Modeli
Bir durağa otobüsler her **$12$ dakikada bir** düzenli olarak gelmektedir. Durağa geliş zamanı tamamen rastgele olan bir yolcunun bekleme süresi $T \sim \mathcal{U}(0, 12)$ dakikadır.
- Yolcunun $4$ dakikadan az bekleme olasılığı nedir?
- Yolcunun $8$ ile $10$ dakika arasında bekleme olasılığı nedir?
- Ortalama bekleme süresi kaç dakikadır?

```python
a_otobus = 0
b_otobus = 12

# 1. P(T <= 4)
p_az_4 = uniform.cdf(4, loc=a_otobus, scale=b_otobus)

# 2. P(8 <= T <= 10)
p_8_10 = uniform.cdf(10, loc=a_otobus, scale=b_otobus) - uniform.cdf(8, loc=a_otobus, scale=b_otobus)

# 3. Ortalama Bekleme Süresi E[T]
ort_bekleme = (a_otobus + b_otobus) / 2

print(f"🚌 4 dakikadan az bekleme olasılığı P(T <= 4)  : %{p_az_4 * 100:.2f} (1/3)")
print(f"🚌 8-10 dakika arası bekleme olasılığı P(8<=T<=10): %{p_8_10 * 100:.2f} (2/12 = 1/6)")
print(f"🚌 Ortalama beklenen süre E[T]                  : {ort_bekleme:.1f} dakika")
```

### 🤖 Vaka 2: Makine Öğrenmesinde Hiperparametre Optimizasyonu (RandomizedSearchCV)
Makine öğrenmesi modellerinde (örn: XGBoost, LightGBM, Derin Öğrenme) öğrenme oranı (learning rate) parametresini ararken sabit grid yerine Düzgün Dağılım üzerinden rastgele örnekleme yapılır.

## 📝 7. Özet ve Hızlı Başvuru Kartı

| Özellik | Sürekli Düzgün $\mathcal{U}(a, b)$ | Ayrık Düzgün $\mathcal{DU}(a, b)$ |
| :--- | :--- | :--- |
| **Tanım Aralığı** | $x \in [a, b]$ (Reel sayılar) | $k \in \{a, a+1, \dots, b\}$ (Tam sayılar) |
| **Olasılık Fonksiyonu** | $f(x) = \frac{1}{b - a}$ (PDF) | $P(X = k) = \frac{1}{n} = \frac{1}{b - a + 1}$ (PMF) |
| **CDF** | $F(x) = \frac{x - a}{b - a}$ | $F(x) = \frac{\lfloor x \rfloor - a + 1}{n}$ |
| **Beklenen Değer $E[X]$** | $\frac{a + b}{2}$ | $\frac{a + b}{2}$ |
| **Varyans $\text{Var}(X)$** | $\frac{(b - a)^2}{12}$ | $\frac{(b - a + 1)^2 - 1}{12}$ |
| **Çarpıklık (Skewness)** | $0$ (Simetrik) | $0$ (Simetrik) |
| **SciPy Karşılığı** | `scipy.stats.uniform(loc=a, scale=b-a)` | `scipy.stats.randint(low=a, high=b+1)` |
| **Kullanım Alanları** | Rastgele Sayı Üreteçleri (RNG), Monte Carlo İntegrali, Bekleme süreleri, Hiperparametre arama |
